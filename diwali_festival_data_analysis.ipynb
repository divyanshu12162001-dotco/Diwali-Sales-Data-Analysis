**Visualization using Seaborn 📈**

```python
 # Gender vs Amount
sns.barplot(x="Gender",y="Amount",data=df,color="orange")
plt.title("sales by gender")
plt.show()
```

```python
# Age Group vs Amount
sns.barplot(x="Age Group",y="Amount",data=df,color="grey")
plt.title("age group sales")
plt.show()
```

```python
# Product Category
plt.figure(figsize=(7,4))
sns.countplot(x="Product_Category",data=df)
plt.xticks(rotation=45)
plt.show()
```

```python
# State-wise sales
state_sales = df.groupby("State")["Amount"].sum().sort_values(ascending=False).head(10)
state_sales
```

```python
sns.barplot(x=state_sales.values,y=state_sales.index ,color="orange")
plt.title("top 10 state sales")
plt.show()
```

```python
# occupation wise sales
occupation_sales = df.groupby("Occupation")["Amount"].sum().sort_values(ascending=True).head(10)
sns.barplot(x=occupation_sales.values,y=occupation_sales.index)
plt.title("top 10 occupation wise sales")
plt.show()
```

```python
# Age + Gender analysis
plt.figure(figsize=(6,4))
sns.countplot(x="Age Group",hue="Gender",data=df,palette="Set2")
plt.xlabel("count")
plt.ylabel("age group")
plt.title("age, gender analysis")
plt.show()
```

**Advance Analysis**

```python
# Age Group + Gender + Amount
age_gender = df.groupby(["Age Group", "Gender"])["Amount"].sum().reset_index()

sns.barplot(
    x="Age Group",
    y="Amount",
    hue="Gender",
    data=age_gender
)

plt.title("Sales by Age Group and Gender")
plt.show()
```

```python
# Occupation + Gender
occupation_gender = df.groupby(["Occupation", "Gender"])["Amount"].sum().reset_index()

sns.barplot(
    x="Occupation",
    y="Amount",
    hue="Gender",
    data=occupation_gender
)

plt.xticks(rotation=45)
plt.show()
```

# Final Insights

Based on the analysis of the Diwali Sales dataset, the following key insights were identified:

* **Gender:** Female customers contributed significantly more to total sales than male customers. Total sales from female customers were approximately ₹7.43 crore, compared with ₹3.19 crore from male customers.
* **Age Group:** The **26–35 age group** generated the highest sales, with approximately ₹4.26 crore in total sales. This group also recorded the highest number of orders.
* **State:** **Uttar Pradesh** generated the highest sales among all states, followed by Maharashtra, Karnataka, and Delhi.
* **Occupation:** Customers working in the **IT Sector** contributed the highest sales among the occupation groups, followed by Healthcare and Aviation.
* **Product Category:** **Food** generated the highest revenue, with approximately ₹3.39 crore in sales.
* **Product Demand:** **Clothing & Apparel** recorded the highest number of orders, followed by Food and Electronics & Gadgets.
* **Customer Segmentation:** The analysis shows that customer characteristics such as gender, age group, state, occupation, and product category can be useful for understanding purchasing patterns.

# Conclusion

The Diwali Sales dataset was analyzed using **Python, Pandas, NumPy, Matplotlib, and Seaborn** to understand customer purchasing behavior and sales patterns. The analysis identified the major customer segments, high-performing states, occupations, and product categories.

The results indicate that the **26–35 age group, female customers, and customers from Uttar Pradesh** were among the strongest contributors to sales. Food was the highest-revenue product category, while Clothing & Apparel had the highest number of orders.

These insights can help businesses better understand their customers, focus marketing campaigns on high-performing customer segments, and improve product and sales strategies.
